# TTR Comparison between Newspaper Articles and Scientific Abstracts

```python
import spacy

# Load the language model
nlp = spacy.load("en_core_web_sm")
```

### TTR Newspaper Articles


```python
#Import libraries 

import pandas as pd
from spacy.lang.en import English

# Load spaCy English model (download if not already installed)
nlp = English()

# Assuming arxiv_data.csv is in your working directory
df = pd.read_csv("cnn4.csv")

# Get the first 20 abstracts and combine them into one string
first_20_news = " ".join(df["SUMMARY"].iloc[:150])
```

```python
#Tokenization

# Create a spaCy document from the combined abstracts
doc = nlp(first_20_news)

# Get the tokens as a list
news_tokens = [token.text.lower() for token in doc if not token.is_stop]

```

```python
# Function to calculate TTR

# Calculate the number of unique tokens (types)
unique_tokens = len(set(news_tokens))

# Calculate TTR (ratio of unique tokens to total tokens)
ttr_news = unique_tokens / len(news_tokens)

print("Newspaper Articles")
print("Type-Token Ratio (TTR):", ttr_news)
```

### TTR Scientific Abstracts


```python
#Import libraries and combine the abstracts

import pandas as pd
from spacy.lang.en import English

# Load spaCy English model
nlp = English()

# Assuming arxiv_data.csv is in your working directory
df2 = pd.read_csv("arxiv_data.csv")

# Get the first 20 abstracts and combine them into one string
first_20_abstracts = " ".join(df2["summaries"].iloc[:150])
```

```python
#Tokenization

# Create a spaCy document from the combined abstracts
doc2 = nlp(first_20_abstracts)

# Get the tokens as a list
sci_tokens = [token.text.lower() for token in doc2 if not token.is_stop]

```

```python
#Calculating TTR

# Calculate the number of unique tokens (types)
unique_tokens = len(set(sci_tokens))

# Calculate TTR (ratio of unique tokens to total tokens)
ttr_sci = unique_tokens / len(sci_tokens)

print("Scientific Abstracts")
print("Type-Token Ratio (TTR):", ttr_sci)
```

```python
import matplotlib.pyplot as plt

# Assuming you have calculated TTR values for scientific abstracts (ttr_sci) and newspaper articles (ttr_news)

# Define labels and data for the bar chart
labels = ["Scientific Abstracts", "Newspaper Articles"]
data = [ttr_sci, ttr_news]

# Create the bar chart
plt.figure(figsize=(8, 6))
plt.bar(labels, data, color=['skyblue', 'lightgreen'])
plt.xlabel("Text Type")
plt.ylabel("Type-Token Ratio (TTR)")
plt.title("TTR Comparison between Scientific Abstracts and Newspaper Articles")

# Display the chart
plt.xticks(rotation=0)  # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Assuming you have calculated ttr_news and ttr_sci

# Define labels for the data
labels = ["Newspaper Articles", "Scientific Abstracts"]

# Create a list containing TTR values
ttr_values = [ttr_news, ttr_sci]

# Create a line chart
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(labels, ttr_values, marker='o', linestyle='-')  # Plot the line with markers

# Set labels and title for the chart
plt.xlabel("Text Type")
plt.ylabel("Type-Token Ratio (TTR)")
plt.title("Comparison of TTR in Newspaper Articles and Scientific Abstracts")

# Add grid lines for better readability
plt.grid(True)

# Show the chart
plt.show()
```
